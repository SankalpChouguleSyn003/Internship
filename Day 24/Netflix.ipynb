```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def load_dataset(): 
    df = pd.read_csv('netflix_titles.csv') 
    print(df)
    return df
df = load_dataset()
```

```python
def summary(df):   
    print(df.head())  
    print(df.info())  
    print(df.dtypes)  
    print(df.describe())  
    print(df.shape)   
    
summary(df)
```

```python
print('Missing values:')  
print(df.isnull().sum())
```

```python
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('NA', inplace=True)
df['country'].fillna(df['country'].mode()[0], inplace=True)
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['date_added'].fillna(df['date_added'].mode()[0], inplace=True)
df['rating'].fillna(df['rating'].mode()[0], inplace=True)
```

```python
print('Missing values after cleaning:')  
print(df.isnull().sum())
```

```python
print('Duplicates')
print(df.duplicated().sum())
```

```python
# Change datatype
df['release_year'] = df['release_year'].astype(int)
```

```python
# To clean duration column
df[['duration_value', 'duration_unit']] = df['duration'].str.extract(r'(\d+)\s*(\w+)')

df['duration_value'] = pd.to_numeric(df['duration_value'], errors='coerce')
```

```python
# Outliers
Q1 = df['duration_value'].quantile(0.25)
Q3 = df['duration_value'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['duration_value'] >= lower) & (df['duration_value'] <= upper)]
```

```python
# Encoding categorical data
df['type_encoded'] = df['type'].map({'Movie': 0, 'TV Show': 1})
```

```python
df.to_csv('cleaned_netflix_dataset.csv', index=False)
```
